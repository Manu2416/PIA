1.	Obtener el nombre y el correo de todos los usuarios

```python
import pandas as pd

df = pd.read_csv("./docsEj/datos_usuarios.csv",encoding="latin1")

df[["name","email"]]
```

2.	Obtener el nombre de todos los usuarios cuyo género es Femenino.

```python
df[df["gender"] == "Femenino"]["name"]
```

3.	Obtener el nombre de todos los usuarios cuyo sexo es Masculino y tengan una edad mayor a 50.

```python
df[(df["gender"] == "Masculino") & (df["age"] > 50)]["name"]
```

4.	Obtener todos los usuarios cuya edad sea mayor a 20

```python
df[df["age"] > 20]
```

5.	Obtener el username y correo electrónico de los usuarios cuya edad se encuentre en el rango 50 a 60.

```python
df[(df["age"] > 50) & (df["age"] < 60)][["username", "age"]]
```

6.	Obtener el username de todos los usuarios cuyo correo empiece con “e”

```python
df[df["username"].str.startswith("e")]
```

7.	Obtener el username de todos los usuarios cuyo país sea Alemania, Finlandia o Canadá

```python

filtro = (df["country"] == "Finland") | (df["country"] == "Germany") | (df["country"] == "Canada")

df[filtro]["name"]
```

8.	Obtener el nombre y correo electronico de todos los usuarios de sexo Femenino que sean del país Alemania

```python
consulta = (df["gender"] == "Femenino") & (df["country"] == "Germany")

df[consulta][["name","email"]]
```

9.	Obtener el promedio de edad de todos los usuarios de sexo Femenino de Canadá mayores de 20 años

```python
filtro = (df["gender"] == "Femenino") & (df["country"] == "Canada") & (df["age"] > 20)
edades = df[filtro]["age"]
promedio = edades.mean()
print(promedio)

```
