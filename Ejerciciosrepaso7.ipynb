6.	(ENTREGA) Dibuja un gráfico para mostrar las temperaturas de todos los días del año agrupadas por meses. Para ello:
a.	Generar un fichero temperaturas txt en el que generes temperaturas de manera aleatoria para los días de todos los meses del año, con variabilidad según las estaciones:
i.	Invierno de 0 a 18 ºC
ii.	Primavera de 10 a 25 ºC
iii.	Verano de 22 a 40 ºC
iv.	Otoño de 15 a 28 ºC
b.	Lee los datos del fichero que acabas de generar y dibuja un gráfico con una línea para cada uno de los meses y sus temperaturas


```python
from random import randint

mescontador = 1
temperaturas = []

def generartemperatura(n1, n2):#funcion que genera todas laas temperaturas
    global mescontador
    mes = [mescontador] 
    for i in range(30):
        temperatura = randint(n1, n2)
        mes.append(temperatura)
    temperaturas.append(mes)
    mescontador += 1  

while True:
    if 1 <= mescontador <= 3:
        generartemperatura(0, 18)
    elif 4 <= mescontador <= 6:
        generartemperatura(10, 25)
    elif 7 <= mescontador <= 9:
        generartemperatura(22, 40)
    elif 10 <= mescontador <= 12:
        generartemperatura(15, 28)
    else:
        break

print(temperaturas)

with open("temperaturas1.txt", "a", encoding="utf-8") as f:
    for lista in temperaturas:
        f.write("\n")  #salto de linea x cada mes
        for i in range(len(lista)):
            f.write(str(lista[i]))
            if i != len(lista) - 1:  #Esto sirve para que no ponga una , en el ultimo valor
                f.write(",")



```

```python
import matplotlib.pyplot as plt


meses = []

with open("temperaturas1.txt", "r", encoding="utf-8") as f:
    lineas = f.readlines()

for linea in lineas:
    linea = linea.strip()   # quita vacios
    if linea:               # ignorar líneas vacías
        mes = []           
        for n in linea.split(","):
            mes.append(int(n))  # convertir a entero
        meses.append(mes)

for mes in meses:
    numero_mes = mes[0]      
    temperaturas = mes[1:]    
    plt.plot(range(1, 31), temperaturas, label=f"Mes {numero_mes}")

plt.xlabel("Dias meses")
plt.ylabel("Temperatura (°C)")
plt.title("Temperaturas diarias por mes")
plt.legend()
plt.show()
```

6.	(ENTREGA) Dibuja un gráfico para mostrar las temperaturas de todos los días del año agrupadas por meses. Para ello:
a.	Generar un fichero temperaturas txt en el que generes temperaturas de manera aleatoria para los días de todos los meses del año, con variabilidad según las estaciones:
i.	Invierno de 0 a 18 ºC
ii.	Primavera de 10 a 25 ºC
iii.	Verano de 22 a 40 ºC
iv.	Otoño de 15 a 28 ºC
b.	Lee los datos del fichero que acabas de generar y dibuja un gráfico con una línea para cada uno de los meses y sus temperaturas

```python
#Otra forma de hacerlo mas facil 

from random import randint
meses = [
    "Enero","Febrero","Marzo","Abril","Mayo","Junio",
    "Julio","Agosto","Septiembre","Octubre","Noviembre","Diciembre"
]
invierno = ["Enero", "Febrero", "Diciembre"]
primavera = ["Marzo", "Abril", "Mayo"]
verano = ["Junio", "Julio", "Agosto"]
otonio  = ["Septiembre", "Octubre", "Noviembre"]



def generar_temperaturas(meses):#funciones que escribe el fichero
     with open("temperaturas2.txt", "a", encoding="utf-8") as f:
        for mes in meses:
            if mes in invierno:
                temp1=0
                temp2=18
            elif mes in primavera:
                temp1=10
                temp2=25
            elif mes in verano:
                temp1=22
                temp2=40
            elif mes in otonio:
                temp1=15
                temp2=28

            for i in range(1, 31):
                        temperatura = randint(temp1,temp2)
                        f.write(f"{mes},{i},{temperatura}\n")                

generar_temperaturas(meses)
           
```

```python
import matplotlib.pyplot as plt

meses = [] 
with open("temperaturas2.txt", "r", encoding="utf-8") as f:
    for linea in f:
        meses.append(linea)

datos_meses = {}#diccionarios para los meses y sus datos

for mes in meses:
    datos = mes.split(",") #lo separamos
    nombre_mes = datos[0]
    dias = int(datos[1])
    temp = int(datos[2])

    if nombre_mes not in datos_meses:
        datos_meses[nombre_mes] = []    #añadimos el nombre del mes si no esta 
    
    datos_meses[nombre_mes].append((dias, temp))    #y sus dias y temps en una lista
    
# print(datos_meses)

for mes, valores in datos_meses.items(): #sacamos el mes y los datos
        dias_lista = [] #creamos una lista para cada uno de los dias y temperaturas
        temp_lista = []
        for dias, temp in valores:
            dias_lista.append(dias) #aqui cargamos los dias y temperaturas
            temp_lista.append(temp)
            
        plt.plot(dias_lista, temp_lista, label=mes)#finalmente los pintamos

plt.xlabel("Días del mes")
plt.ylabel("Temperatura (°C)")
plt.title("Temperaturas diarias por mes")
plt.legend()
plt.show()

```
